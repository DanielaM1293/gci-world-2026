```python
import pandas as pd

def tarea(datos_anime_extraídos):
  # Agrupamos por 'Tipo', seleccionamos 'Puntuación', calculamos la media
  # y ordenamos los valores de forma descendente (ascendente = False)
  resultado = (
      datos_anime_extraídos.groupby('Tipo')['Puntuación'].mean().sort_values(
          ascending=False
      )
  )

  return resultado
```
