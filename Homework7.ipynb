```python
import pandas as pd

def tarea(datos_anime_extraídos):
    # 1. Crear la columna de deciles usando qcut
    datos_anime_extraídos['Members_decile'] = pd.qcut(
        datos_anime_extraídos['Miembros'],
        q=10,
        labels=['01', '02', '03', '04', '05', '06', '07', '08', '09', '10'],
        observed=True
    )

    # 2. Agrupar y calcular las agregaciones requeridas
    resultado = datos_anime_extraídos.groupby('Members_decile', observed=True).agg(
        Puntuación_media=('Puntuación', 'mean'),
        Total_de_miembros=('Miembros', 'sum')
    )

    # 3. Asignar nombre al índice
    resultado.index.name = 'Miembros'

    # 4. Calcular el porcentaje acumulado usando cumsum()
    total = resultado['Total_de_miembros'].sum()
    resultado['Porcentaje_acumulado'] = (resultado['Total_de_miembros'].cumsum() / total) * 100

    # 5. Seleccionar y retornar las columnas exactas ordenadas
    return resultado[['Puntuación_media', 'Total_de_miembros', 'Porcentaje_acumulado']]
```
