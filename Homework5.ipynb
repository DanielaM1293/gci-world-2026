```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score

def tarea(X_train, y_train, tipo_modelo, k):
    # 1. Selección del modelo con random_state=0 según el tipo recibido
    if tipo_modelo == 'logistic_regression':
        clf = LogisticRegression(random_state=0)
    elif tipo_modelo == 'decision tree':
        clf = DecisionTreeClassifier(random_state=0)
    elif tipo_modelo == 'svm':
        clf = SVC(random_state=0)

    # 2. Creación de la canalización (Pipeline)
    pipeline = Pipeline([
        ('escalador', StandardScaler()),
        ('clasificador', clf)
    ])

    # 3. Evaluación con validación cruzada usando cross_val_score
    puntuaciones = cross_val_score(pipeline, X_train, y_train, cv=k)

    # 4. Retornamos el promedio de las puntuaciones convertido a float
    return float(puntuaciones.mean())
```
