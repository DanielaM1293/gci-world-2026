```python
import pandas as pd

def tarea_corregida(datos_anime, columna_métrica, n):
  # 1. Ordenar y hacer una copia
  anime_sorted = datos_anime.sort_values(
      by=columna_métrica, ascending=False
  ).copy()

  # 2. Calcular los rangos usando el método 'first'
  ranks = anime_sorted[columna_métrica].rank(method='first', ascending=False)

  # 3. Dividir en n grupos usando qcut con base en los rangos
  # Obtenemos IDs de grupo del 0 al n-1
  anime_sorted['group_id'] = pd.qcut(ranks, q=n, labels=False)

  # 4. Calcular la suma total de la columna métrica
  suma_total = anime_sorted[columna_métrica].sum()

  # 5. Agrupar, sumar por grupo y calcular el porcentaje respecto al total
  # El índice de 'resultado' serán los group_ids (0, 1, ..., n-1)
  resultado = (
      anime_sorted.groupby('group_id')[columna_métrica].sum() / suma_total
  )

  # 6. Renombrar las etiquetas del índice de forma segura
  # Usamos una comprensión de lista para generar etiquetas basadas en los índices actuales.
  # El '+1' es porque los group_ids empiezan en 0 y queremos etiquetas del 1 al n.
  resultado.index = [f'Grupo {i + 1}' for i in resultado.index]

  # 7. Ordenar los resultados finales de mayor a menor porcentaje
  # Ahora ordenamos por los valores de porcentaje, no por las etiquetas del índice.
  resultado = resultado.sort_values(ascending=False)

  return resultado
```
