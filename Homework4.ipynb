```python
from sklearn.linear_model import LinearRegression


def tarea(datos_anime_extraídos, columna_X, columna_Y):
  # 1. Inicializamos el modelo de Regresión Lineal
  modelo = LinearRegression()

  # 2. Preparamos X como matriz bidimensional y Y como vector usando .values
  X = datos_anime_extraídos[[columna_X]].values
  Y = datos_anime_extraídos[columna_Y].values

  # 3. Entrenamos el modelo con los datos usando .fit()
  modelo.fit(X, Y)

  # 4. Calculamos el coeficiente de determinación R^2 usando .score()
  resultado = modelo.score(X, Y)

  # Retornamos el resultado convertido a tipo float estándar de Python
  return float(resultado)
```
