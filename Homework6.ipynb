```python
import pandas as pd

def tarea(datos_anime_extraídos):
  # 1. Estandarizar la columna Puntuación (media=0, desviación estándar=1)
  media_puntuacion = datos_anime_extraídos['Puntuación'].mean()
  puntuacion_std = datos_anime_extraídos['Puntuación'].std()

  puntuacion_estandarizada = (datos_anime_extraídos['Puntuación'] - media_puntuacion) / puntuacion_std

  # 2. Codificación one-hot de la columna 'Tipo' usando los datos correctos de la función
  type_dummies = pd.get_dummies(datos_anime_extraídos['Tipo'], dtype=int)

  # 3. Asegurar que todas las columnas requeridas estén presentes (rellenando con 0 si faltan)
  columnas_requeridas = ['Película', 'Música', 'ONA', 'OVA', 'Especial', 'TV']
  for col in columnas_requeridas:
    if col not in type_dummies.columns:
      type_dummies[col] = 0

  # 4. Seleccionar y ordenar las columnas requeridas en el orden exacto
  type_dummies = type_dummies[columnas_requeridas]

  # 5. Combinar la puntuación estandarizada con las columnas one-hot de Type
  resultado = pd.concat(
      [
          puntuacion_estandarizada.rename('Score_std').reset_index(drop=True),
          type_dummies.reset_index(drop=True)
      ],
      axis=1
  )

  # 6. Retornar el resultado final
  return resultado
```
